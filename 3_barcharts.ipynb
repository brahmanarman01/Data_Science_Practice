```python
# Import the matplotlib.pyplot library for plotting
import matplotlib.pyplot as plt
```

```python
# Define lists for years and runs scored
years=[1,2,3,4,5,6,7,8]
runs=[8,7,6,5,4,3,2,1]
```

```python
# Create a bar plot of Sachin Tendulkar's yearly runs
plt.bar(years,runs)
plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Sachin Tendulkar's Yearly Runs")
plt.show()
```

```python
# Import the numpy library
import numpy as np

# Define lists for runs scored by different players over the years
sachin = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
sehwag = [0, 200, 900, 1400, 1600, 1800, 1500, 1100, 800, 0]
kohli  = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]

# Create an array of index positions for the years
x=np.arange(len(years))#  give index position of each value more abou this np.arange([start,] stop[, step], dtype=None)
x

# Define the width of the bars for the grouped bar chart
width=0.25
```

```python
# Create a grouped bar chart comparing runs of different players over the years
from pathlib import WindowsPath

plt.bar(x-width,sehwag,width=width,label="sachin")
plt.bar(x,sehwag,width=width,label="sehwag")
plt.bar(x+width,sehwag,width=width,label="kohli")
plt.xlabel("Year")
plt.ylabel("Runs")
plt.title("Run Comparison")
plt.xticks(x, years)  # Show actual year instead of 0,1,2,...
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Create a bar chart showing runs scored by specific players and add text labels on top
import matplotlib.pyplot as plt

players = ["Sachin", "Sehwag", "Kohli"]
runs = [1500, 1200, 1800]

plt.bar(players, runs, color="skyblue")

# Add labels on top of bars
for i in range(len(players)):
    plt.text(i, runs[i] + 50, str(runs[i]), ha='center')

plt.ylabel("Runs")
plt.title("Runs Scored by Players")
plt.tight_layout()
plt.show()
```
