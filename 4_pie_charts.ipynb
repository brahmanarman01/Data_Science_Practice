```python
import matplotlib.pyplot as plt
```

```python
plt.style.use("ggplot")
```

```python
# Data for the pie chart
labels = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs = [18000, 8000, 12000, 9500]

# Create a pie chart
plt.title("Career Runs of Indian Batsmen")
plt.pie(runs, labels=labels)
plt.show()
```

```python
# Define custom colors for the pie chart slices
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99']
plt.pie(runs, labels=labels, colors=colors,
         wedgeprops={'edgecolor': 'black', 'linewidth': 2, 'linestyle': '--'})
plt.show()
```

```python
# Define explode values to separate slices
explode = [0.1, 0, 0.2, 0]

# Create an exploded pie chart with shadow and rotated start angle
plt.pie(runs, labels=labels, explode=explode, colors=colors,  shadow=True, startangle=140 )
plt.title("Exploded Pie Example")
plt.show()
```

```python
# if there are big no of categories then you must to be avoid plotting a pie
# chart

# Data for a pie chart with many categories
categories = [
    "Category 1", "Category 2", "Category 3", "Category 4", "Category 5",
    "Category 6", "Category 7", "Category 8", "Category 9", "Category 10",
    "Category 11", "Category 12", "Category 13", "Category 14", "Category 15",
    "Category 16", "Category 17", "Category 18", "Category 19", "Category 20"
]

data = [
    450, 320, 210, 560, 710,
    390, 275, 600, 150, 490,
    380, 420, 310, 505, 660,
    240, 330, 410, 295, 530
]
# Define custom colors for the pie chart slices
colors = [
    "#FF5733", "#33FF57", "#3357FF", "#F39C12", "#9B59B6",
    "#1ABC9C", "#E74C3C", "#2ECC71", "#3498DB", "#E67E22",
    "#8E44AD", "#16A085", "#D35400", "#27AE60", "#2980B9",
    "#C0392B", "#7D3C98", "#117A65", "#CA6F1E", "#2E86C1"
]
# Define explode values for each slice
explode=[0.1]*20
plt.figure(figsize=(8, 8))
# Create a pie chart with many categories, exploded slices, and percentages
plt.pie(data, labels=categories, autopct='%1.1f%%',colors=colors,explode=explode,shadow=True)
plt.title("Pie Chart with Exploded Slices",fontsize=16)

plt.show()
```
