```python
import requests
# This cell imports the requests library, which is used to make HTTP requests.
```

```python
a=requests.get("https://books.toscrape.com/")
# This cell makes a GET request to the specified URL and stores the response in variable 'a'.
```

```python
# This cell is currently empty and can be used for further code.
```

```python
# This cell iterates through pages 1 to 50 of the website,
# downloads the HTML content of each page, and saves it to a file in the 'htmls' directory.
for i in range(1, 51):
  a=requests.get(f"https://books.toscrape.com/catalogue/page-{i}.html")
  with open(f"htmls/page{i}.html", "w",encoding="utf-8") as f:
    f.write(a.text)
    print(f"page{i} succesfylly download")
    print(a.status_code)
```
