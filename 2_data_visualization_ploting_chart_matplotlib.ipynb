```python
# Import the matplotlib library for plotting
import matplotlib.pyplot as plt
```

```python
# Create a simple line plot with default x and y values
plt.plot([1, 2, 3], [4, 5, 6])
# Display the plot
plt.show()
```

```python
# Define data for years and runs
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]

# Create a line plot of runs over the years
plt.plot(years, runs)
# Display the plot
plt.show()
```

```python
# Set labels for the x and y axes and a title for the plot
plt.xlabel("year")
plt.ylabel("runs")
plt.title("kl rahul runs per year")
# Create a line plot of runs over the years
plt.plot(years, runs)
# Display the plot
plt.show()
```

```python
# Define data for multiple players' runs over the years
kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]
dhoni = [1, 200, 300, 1100, 1600, 1800, 1500, 1540, 1000, 0]
# Plot the runs for each player with different markers and line styles
plt.plot(years, kohli,"ro--",label="kohli")
plt.plot(years, sehwag,"g^", label="sehwag")
plt.plot(years, dhoni,"bD", label="dhoni")
# Add a grid to the plot
plt.grid('true')
# Adjust layout to prevent labels overlapping
plt.tight_layout()
# Add a legend to identify each line
plt.legend()

# Display the plot
plt.show()
```

```python
# Apply the 'ggplot' style to plots
plt.style.use("ggplot")
```

```python
# Use the 'xkcd' style for plotting (often results in a hand-drawn look)
with plt.xkcd():
    # Plot data for Kohli and Sehwag
    plt.plot(years, kohli, label="Kohli")
    plt.plot(years, sehwag, label="Sehwag")
    # Add a title to the plot
    plt.title("Epic Battle of the Batsmen")
    # Add a legend to identify each line
    plt.legend()
    # Display the plot
    plt.show()
```

```python
# Import the numpy library for numerical operations
import numpy as np

# Generate and plot 50 random lines
for i in range(50):
    plt.plot(np.random.rand(100), linewidth=1)

# Add a title to the plot
plt.title("Too Much Data Can Be Confusing!")
# Add a grid to the plot
plt.grid(True)
# Adjust layout to prevent labels overlapping
plt.tight_layout()
# Display the plot
plt.show()
```
