```python
# Import the matplotlib library for plotting
import matplotlib.pyplot as plt
```

```python
# Define the activities and the time spent on each activity in hours
activities = ['Studying', 'Playing', 'Watching TV', 'Sleeping']
time_spent = [3, 2, 2, 5]  # hours in a day
```

```python
# Define colors for each activity in the pie chart
colors = ['skyblue', 'lightgreen', 'gold', 'lightcoral']
```

```python
# Create a pie chart to visualize the distribution of time spent on after-school activities
plt.figure(figsize=(3,3))
plt.pie(time_spent, labels=activities, colors=colors, autopct='%1.1f%%', startangle=90)
plt.title("After School Activities")
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

```python
# Import the matplotlib library for plotting
import matplotlib.pyplot as plt

# Define the days of the week and the time spent on different activities each day
days = [1, 2, 3, 4, 5, 6, 7]  # Days of the week
studying = [3, 4, 3, 5, 4, 3, 4]
playing = [2, 2, 1, 1, 2, 3, 2]
watching_tv = [2, 1, 2, 2, 1, 1, 1]
sleeping = [5, 5, 6, 5, 6, 5, 5]

# Define labels and colors for the stacked area plot
labels = ['Studying', 'Playing', 'Watching TV', 'Sleeping']
colors = ['skyblue', 'lightgreen', 'gold', 'lightcoral']

# Create a stacked area plot to visualize the weekly activity tracker
plt.figure(figsize=(10,6))
plt.stackplot(days, studying, playing, watching_tv, sleeping,
              labels=labels, colors=colors, alpha=0.8)

# Add a legend, title, labels for x and y axes, and a grid to the plot
plt.legend(loc='upper left')  # Location of the legend
plt.title('Weekly Activity Tracker')
plt.xlabel('Day')
plt.ylabel('Hours')
plt.grid(True)
plt.show()
```
